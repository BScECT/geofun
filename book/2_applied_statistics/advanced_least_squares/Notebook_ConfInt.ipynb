## Observation model

The code cell below sets up our problem in three steps:
1. Set the observation model using the time when the observations were taken (epoch) to create our $\boldsymbol{X}$-matrix
2. Get samples of our random observations, `y`, with error `e`

The simulated set of observations uses the `random.seed` method to ensure that the realizations in the random sample are always the same. This is to ensure that everyone gets the same result; it is still a random sample. 

````python
t = np.vstack([1,3,3.5,5,5.5,7,9,10,10.5,
               11.5,13,14,16,16.5,18,19,
               20,22.5,24,25])
n = len(t)
X = np.hstack((np.ones((n,1)),t))
sigma = 2

np.random.seed(1613353294)
e = np.random.normal(0, sigma, n)
xo = [3,0.3]
y = X @ xo + e
````

````python
n_pred = 30
t_pred = np.arange(1,n_pred+1)
X_pred = np.column_stack((np.ones(n_pred), t_pred))
````

````python
def BLUE(X, y, Sigma_Y):
    """ 
    Function to calculate the Best Linear Unbiased Estimator
    Output: 
    betahat        estimated parameters
    Sigma_betahat  covariance matrix of estimated parameters
    yhat           predicted observations
    """
    W = np.linalg.inv(Sigma_Y)
    Sigma_betahat = np.linalg.inv(X.T @ W @ X)
    betahat = Sigma_betahat @ X.T @ W @ y
    yhat = X @ betahat
    return betahat, Sigma_betahat, yhat
````

````python
def BLUE_predict(X_pred, betahat, Sigma_Xhat):
    """ 
    Function to calculate the fitted model (yhat) incl. predictions
    Output: 
    yhat_pred        predicted observations
    Sigma_Yhat_pred  covariance matrix of predicted observations
    """
    yhat_pred = X_pred @ betahat
    Sigma_Yhat_pred = X_pred @ Sigma_Xhat @ X_pred.T
    return yhat_pred, Sigma_Yhat_pred
````

````python
def conf_interval(Sigma_Yhat, conf_level):
    """ 
    Function to calculate confidence interval of fitted model
    conf_level is the confidence level as percentage (e.g., 95)
    Output: 
    CI_yhat  confidence bound of yhat 
    k        CI_yhat[i] = k * sigma_yhat[i]
    """
    alpha = 1 - conf_level/100
    k = norm.ppf(1-0.5*alpha)
    CI_yhat = k * np.sqrt(np.diagonal(Sigma_Yhat))
    return CI_yhat, k
````

````python
def plot_all(t, y, X, t_pred, X_pred, sigma, conf_level):
    """ 
    Function to create plot with observations, fitted model 
    and confidence bounds
    """
    n = len(t)
    Sigma_Y = sigma**2*np.eye(n) 
    betahat, Sigma_betahat, yhat = BLUE(X, y, Sigma_Y)
    yhat_pred, Sigma_Yhat_pred = BLUE_predict(X_pred, betahat, Sigma_betahat)
    CI_yhat, k = conf_interval(Sigma_Yhat_pred, conf_level)
    
    # create plot with observations, error bars and confidence intervals
    plt.figure(figsize = (10,6))
    plt.xlabel('t [s]')
    plt.ylabel('x(t) [m]')

    plt.plot(t, y, 'k*',
             label=f'observations with {conf_level:.1f}% conf.')
    plt.title('$n$='+str(n)+' and $\sigma$='+str(sigma)+' [m]')

    # plot observations with errorbars   
    plt.errorbar(t, y, yerr = k*sigma,
                 fmt='', capsize=5, linestyle='')

    # plot model and predicted observations
    plt.plot(t, yhat, 'go')

    # plot confidence intervals
    plt.plot(t_pred, yhat_pred + CI_yhat, 'r',
             label=f'{conf_level:.1f}% conf.')
    plt.plot(t_pred, yhat_pred - CI_yhat, 'r')
    plt.plot(t_pred, yhat_pred, 'g')
    plt.ylim(-2, 20)
    plt.legend()
````

## Plot observations and 95% confidence intervals with all observations
You can also try different confidence levels.

````python
# default model
plot_all(t, y, X, t_pred, X_pred, sigma, 95)
````

## What if we reduce the number of observations (only the central 10)

:::{card}

Compare with the first figure and discuss the differences.

 ```{admonition} Discussion
* Since we have a different set of observations, the estimates will be different.
* With fewer observations, the precision will be worse, hence the wider confidence intervals.
* Since all observations are in the centre, there is much more uncertainty in the values outside the range of observation times.
 ```


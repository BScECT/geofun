# Regression & least squares

## What is regression?

Regression predicts an outcome from a set of predictors using a mathematical model. Its two main uses are:
  * For understanding the relationships between variables, which is called regression analysis in statistics.
  * For prediction and forcasting, which is more and more linked to machine learning.

In general, the mathematical model behind regression takes the form:

$$y = f(x,\beta) + \epsilon$$

Or, in words:

$$\text{outcome} = \text{function}(\text{predictors},\text{parameters}) + \text{error}$$

With:

  * *Outcome*: the variable to predict, available from the data; depending on the field, it is also known as output, target, label, response variable, dependent variable, or endogenous variable.
  * *Predictors*: the variables to guide the predictions, available from the data as well; depending on the field, they are also known as inputs, features, regressors, covariates, explanatory variables, independent variables, or exogenous variables.
  * *Parameters*: unknown constants that control the relationship between predictors and outcome and must be estimated from the data; depending on the field, they are also known as coefficients, weights, or effects.
  * *Error*: measurement and model error, i.e., variations unexplained by the model; depending on the field, it is also known as noise or disturbance.

The function and its parameters are an essential component of the regression model. They must be chosen depending on the patterns between predictors and outcome identified during exploratory data analysis. In this section, we will start with the simplest model there is to predict continuous variables: the linear model.

(2:regression:simple)=
### Simple linear regression

Simple linear regression deals with a single predictor in a linear model:

$$y = \beta_0 + \beta_1 x + \epsilon$$

It has two parameters ({numref}`figure {number} <linear_model_params>`):

  * The intercept $\beta_0$, which is the value of the outcome $y$ when the predictor $x$ is 0.
  * The slope $\beta_1 = \displaystyle\frac{\text{change in }y}{\text{change in }x}$, which defines how changes in $x$ impact $y$.

(2:regression:multiple)=
### Multiple linear regression

Multiple linear regression deals with two or more predictors, still in a linear model:

$$y = \beta_0 + \beta_1 x_1 + \cdots + \beta_m x_m + \epsilon$$

It has $m + 1$ parameters, where $m$ is the number of predictors: an intercept, and one slope per predictor.

```{admonition} Link to linear algebra
:class: tip

When used on a dataset, the formula above needs to be applied to each of the $n$ observations:

$$y_i = \beta_0 + \beta_1 x_{1,i} + \cdots + \beta_m x_{m,i} + \epsilon_i \qquad \forall i \in \{1, \dots, n\}$$

In practice, it is more efficient to use its matrix form:

$$
\begin{bmatrix}
    y_1 \\
    y_2 \\
    \vdots \\
    y_n \\
\end{bmatrix}
=
\begin{bmatrix}
    1 & x_{1,1} & \cdots & x_{m,1} \\
    1 & x_{1,2} & \cdots & x_{m,2} \\
    \vdots & \vdots & \vdots & \vdots \\
    1 & x_{1,n} & \cdots & x_{m,n} \\
\end{bmatrix}
\begin{bmatrix}
    \beta_0 \\
    \beta_1 \\
    \vdots \\
    \beta_m \\
\end{bmatrix}
+
\begin{bmatrix}
    \epsilon_0 \\
    \epsilon_1 \\
    \vdots \\
    \epsilon_m \\
\end{bmatrix}
$$

Or in condensed notation:

$$\boldsymbol{y} = \boldsymbol{X}\boldsymbol{\beta} + \boldsymbol{\epsilon}$$

Where:
  * $\boldsymbol{y}$ is called the observation vector and includes the outcome.
  * $\boldsymbol{X}$ is called the design matrix and includes the predictors with a column of 1 for the intercept.
  * $\boldsymbol{\beta}$ is called the parameter vector.
  * $\boldsymbol{\epsilon}$ is called the error vector.
```

## What is least squares?

The parameters being unknown, we first need to estimate them from the data before being able to use a regression model ({numref}`figure {number} <linear_model_fit>`). This is generally called fitting the model to the data, although depending on the field it is also known as training, callibration, or estimation.

````{iframe-figure} ../../_static/part-b_linear-model_fit.html

:width: 880px

````

Least squares refers to a set of methods that determine the best-fit parameter values by minimizing the sum of the squared residuals. This can be done through iterative optimization procedures, and we will talk more about that later. But linear models are simple enough that we can derive a closed-form solution to get the best parameter values for a given dataset.

(2:regression:residuals)=
### Residuals

To make predictions with a linear model, we need estimates of the intercept and slope, $\hat{\beta}_0$ and $\hat{\beta}_1$, so that we can get estimates of the outcome, $\hat{y}$:

$$\hat{y} = \hat{\beta}_0 + \hat{\beta}_1 x$$

From there, we can also get estimates of the error, called the residuals, by comparing the predictions to the data:

$$
    \begin{align}
        r = \hat{\epsilon} &= y - \hat{y} \\
         &= y - \hat{\beta}_0 - \hat{\beta}_1 x \\
    \end{align}
$$

The sum of the squares of the residuals gives us a single value quantifying how well a model and its current parameter values match the data:

$$
    \begin{align}
        S_{\text{res}}(\hat{\beta}_0,\hat{\beta}_1) &= \sum_{i=1}^n r_i^2 \\
         &= \sum_{i=1}^n (y_i - \hat{y}_i)^2 \\
         &= \sum_{i=1}^n (y_i - \hat{\beta}_0 - \hat{\beta}_1 x_i)^2 \\
    \end{align}
$$

We can then use this function to find the set of parameter values that best fit the data.

(2:regression:ols)=
### Ordinary least squares

Finding those parameter values means solving the following minimization problem:

$$\underset{\hat{\beta}_0,\hat{\beta}_1}{\min} S_{\text{res}}(\hat{\beta}_0,\hat{\beta}_1)$$

To find a closed-form solution to that problem, we need to solve:

$$\frac{\partial S_{\text{res}}}{\partial \hat{\beta}_0} = 0; \qquad \frac{\partial S_{\text{res}}}{\partial \hat{\beta}_1} = 0$$

Let's start with the first one and find $\hat{\beta}_0$. First, we apply the linearity of differenciation:

$$
    \begin{align}
        \frac{\partial S_{\text{res}}}{\partial \hat{\beta}_0} &= \frac{\partial}{\partial \hat{\beta}_0}\sum_{i=1}^n r_i^2 \\
        &= \sum_{i=1}^n\frac{\partial r_i^2}{\partial \hat{\beta}_0} \\
    \end{align}
$$

Then the chain rule:

$$
    \begin{align}
        \frac{\partial S_{\text{res}}}{\partial \hat{\beta}_0} &= \sum_{i=1}^n\frac{\partial r_i^2}{\partial r_i}\frac{\partial r_i}{\partial \hat{\beta}_0} \\
        &= \sum_{i=1}^n 2r_i \cdot (-1) \\
        &= -2\sum_{i=1}^n (y_i - \hat{\beta}_0 - \hat{\beta}_1 x_i) \\
    \end{align}
$$

From there, remember that the sample mean for $x$ and $y$ are:

$$
    \begin{align}
        \overline{x} &= \frac{1}{n}\sum_{i=1}^n x_i \\
        \overline{y} &= \frac{1}{n}\sum_{i=1}^n y_i \\
    \end{align}
$$

And we can use those to remove the sum:

$$
    \frac{\partial S_{\text{res}}}{\partial \hat{\beta}_0} = -2n(\overline{y} - \hat{\beta}_0 - \hat{\beta}_1 \overline{x}) = 0
$$

So that:

$$\hat{\beta}_0 = \overline{y} - \hat{\beta}_1 \overline{x}$$

Now we have the intercept, which depends on the slope. So let's turn to the second of our initial partial derivatives to find $\hat{\beta}_1$. First, we can substitute for $\hat{\beta}_0$ in the sum of the squares of the residuals:

$$
    \begin{align}
        S_{\text{res}}(\hat{\beta}_0,\hat{\beta}_1) &= \sum_{i=1}^n (y_i - \hat{\beta}_0 - \hat{\beta}_1 x_i)^2 \\
         &= \sum_{i=1}^n (y_i - \overline{y} + \hat{\beta}_1 \overline{x} - \hat{\beta}_1 x_i)^2 \\
         &= \sum_{i=1}^n \left[(y_i - \overline{y}) - \hat{\beta}_1 (x_i - \overline{x})\right]^2 \\
    \end{align}
$$

Let's call the element within the sum $r'_i$:

$$r'_i = (y_i - \overline{y}) - \hat{\beta}_1 (x_i - \overline{x})$$

Again, we apply the linearity of differenciation and the chain rule:

$$
    \begin{align}
        \frac{\partial S_{\text{res}}}{\partial \hat{\beta}_1} &= \sum_{i=1}^n\frac{\partial {r'}_i^2}{\partial r'_i}\frac{\partial r'_i}{\partial \hat{\beta}_1} \\
        &= \sum_{i=1}^n 2r'_i \cdot (\overline{x} - x_i) \\
        &= -2\sum_{i=1}^n \left[(y_i - \overline{y}) - \hat{\beta}_1 (x_i - \overline{x})\right](x_i - \overline{x}) \\
        &= -2\left(\sum_{i=1}^n (y_i - \overline{y})(x_i - \overline{x}) - \hat{\beta}_1\sum_{i=1}^n (x_i - \overline{x})^2 \right) \\
        &= 0 \\
    \end{align}
$$

So that:

$$
    \hat{\beta}_1 = \frac{\sum_{i=1}^n (y_i - \overline{y})(x_i - \overline{x})}{\sum_{i=1}^n (x_i - \overline{x})^2}
$$

And we now have the two formulas to get the ordinary least squares estimates for simple linear regression:

$$
    \begin{align}
        \hat{\beta}_1 &= \frac{\sum_{i=1}^n (y_i - \overline{y})(x_i - \overline{x})}{\sum_{i=1}^n (x_i - \overline{x})^2} \\
        \hat{\beta}_0 &= \overline{y} - \hat{\beta}_1 \overline{x} \\
    \end{align}
$$

Similar steps can be followed to find estimates for the multiple linear regression.

```{admonition} Link to linear algebra
:class: tip

Residuals and sum of the squares of the residuals can also be written in matrix form:

$$
    \begin{align}
        \boldsymbol{r} &= \boldsymbol{y} - \boldsymbol{X}\boldsymbol{\hat{\beta}} \\
        S_{\text{res}}(\boldsymbol{\hat{\beta}}) &= \boldsymbol{r}^T \boldsymbol{r} \\
    \end{align}
$$

And, similarly, we can define a closed-form solution to the least squares problem in matrix form, starting with substituting the residuals in $S_{\text{res}}$ with their equations for a linear model:

$$
    \begin{align}
        S_{\text{res}}(\boldsymbol{\hat{\beta}}) &= \left(\boldsymbol{y} - \boldsymbol{X}\boldsymbol{\hat{\beta}}\right)^T \left(\boldsymbol{y} - \boldsymbol{X}\boldsymbol{\hat{\beta}}\right) \\
         &= \boldsymbol{y}^T \boldsymbol{y} - \boldsymbol{y}^T \boldsymbol{X}\boldsymbol{\hat{\beta}} - \boldsymbol{\hat{\beta}}^T \boldsymbol{X}^T \boldsymbol{y} + \boldsymbol{\hat{\beta}}^T \boldsymbol{X}^T \boldsymbol{X}\boldsymbol{\hat{\beta}} \\
    \end{align}
$$

Now, keep in mind that $\boldsymbol{\hat{\beta}}^T$ is a vector of dimensions $(1 \times m + 1)$, $\boldsymbol{X}^T$ is a matrix of dimensions $(m + 1 \times n)$, and $\boldsymbol{y}$ is a vector of dimensions $(n \times 1)$, so that applying matrix multiplication returns a scalar:

$$(1 \times m + 1) \cdot (m + 1 \times n) \cdot (n \times 1) = (1 \times 1)$$

So we have:

$$\boldsymbol{y}^T \boldsymbol{X}\boldsymbol{\hat{\beta}} = \left(\boldsymbol{y}^T \boldsymbol{X}\boldsymbol{\hat{\beta}}\right)^T = \boldsymbol{\hat{\beta}}^T \boldsymbol{X}^T \boldsymbol{y}$$

Which leads to:

$$S_{\text{res}}(\boldsymbol{\hat{\beta}}) = \boldsymbol{y}^T \boldsymbol{y} - 2\boldsymbol{\hat{\beta}}^T \boldsymbol{X}^T \boldsymbol{y} + \boldsymbol{\hat{\beta}}^T \boldsymbol{X}^T \boldsymbol{X}\boldsymbol{\hat{\beta}}$$

From there, we derive again the sum of the squares of the residuals with the parameters and find when it is equal to zero (i.e., we find the minimum of the sum of the squares of the residuals):

$$\frac{\partial S_{\text{res}}}{\partial \boldsymbol{\hat{\beta}}} = \frac{\partial}{\partial \boldsymbol{\hat{\beta}}}\left( \boldsymbol{y}^T \boldsymbol{y} - 2\boldsymbol{\hat{\beta}}^T \boldsymbol{X}^T \boldsymbol{y} + \boldsymbol{\hat{\beta}}^T \boldsymbol{X}^T \boldsymbol{X}\boldsymbol{\hat{\beta}} \right) = -2\boldsymbol{X}^T \boldsymbol{y} + 2\boldsymbol{X}^T \boldsymbol{X}\boldsymbol{\hat{\beta}} = 0$$

Which gives us the so-called normal equations:

$$\boldsymbol{X}^T \boldsymbol{X}\boldsymbol{\hat{\beta}} = \boldsymbol{X}^T \boldsymbol{y}$$

So that:

$$\boldsymbol{\hat{\beta}} = \left(\boldsymbol{X}^T \boldsymbol{X}\right)^{-1} \boldsymbol{X}^T \boldsymbol{y}$$

Here we directly get estimates for all the parameters with a single formula, in the simple and multiple cases, but the inversion of $\boldsymbol{X}^T \boldsymbol{X}$ can be unstable.
```

(2:regression:assumptions)=
### Assumptions

Linear regression with ordinary least squares relies on a set of assumptions:

  * *Linearity*: a linear relation exists between predictors and outcome.
  * *No multicolinearity*: the predictors are not linearly related.
  * *Noncorrelation*: the predictors are uncorrelated with the errors.
  * *Centering*: the errors have a mean of 0; some operations also require them to be normally distributed.
  * *Homoscedasticity*: the errors have a constant variance.
  * *Independence*: the errors are independent (so not autocorrelated, i.e., they do not follow some specific patterns).

When those assumptions are met, it is the best linear unbiased estimator (nicknamed BLUE): the parameter estimates are equal to the true parameter values (*unbiased*) and have minimal variance (*best*), i.e., they have the smallest uncertainty due to having a finite sample.

## How to interpret a linear model?

(2:regression:interpretation)=
### Parameter interpretation

As mentioned at the beginning, the intercept is the value of the outcome when the predictors are equal to 0, and the slope quantifies by how much the outcome changes when a predictor changes. Let's see what that means by looking at the association between porosity and density on [some core samples from Mt. Simon Sandstone](https://www.ideals.illinois.edu/items/119677):
  * The Pearson correlation coefficient between porosity and density is -0.62, so we have a moderate negative association between the two variables.

  * The linear regression with porosity as predictor and density as outcome leads to an intercept of 2.6613 and a slope of -0.0018, so:
    * When porosity is null, density is equal to 2.6613 g/cm$^3$, which is consistent with [mineralogical observations of a dominance of quartz grains](https://hdl.handle.net/11299/212483).
    * When porosity increases by 1%, density decreases by 0.0018 g/cm$^3$.
    So regression analysis gives us more information than the correlation coefficient.

  * The same regression analysis the other way around, with density as predictor and porosity as outcome, leads to an intercept of 566.7 and a slope of -209.8, so:
    * When density is null, porosity is equal to 566.7%.
    * When density increases by 1 g/cm$^3$, porosity decreases by 209.8%.

    Here, the intercept has lost any physical meaning: porosity cannot be higher than 100%, and that would mean no core sample anymore (same for a density of 0). This is because density depends on porosity, not the other way around: [cause-and-effect relationships](1:descriptive_statistics:causation) matter in regression analysis.

In the end, the correlation coefficient quantifies the strength of an association between two variables, but not how one variable affects the other. Regression analysis can do the later. Multiple linear regression can even go one step further: since it quantifies the impact of multiple predictors together, the slope of one predictor gives us its effect on the outcome adjusted for the effect of all other predictors. For instance, adding water saturation as a predictor leads to an intercept of 2.659, a slope of -0.0021 for the porosity, and a slope of 0.0001 for the water saturation. So now the slope for the porosity only includes the effect of porosity on density, and not the effect of water in the pores.

(2:regression:determination)=
### Coefficient of determination

The coefficient of determination is defined as the proportion of the outcome's variation that is explained by the model, so it a measure of the goodness of fit of a model. The closer it gets to 1, the more variability of the data is captured by the model (which is not always a good thing, we will see that later when discussing validation). To define it, we need three quantities:
  * The deviation between the outcome's mean and the data, called total deviation from the mean and also designated $TSS$:

    $$
        S_{\text{tot}}(\hat{\beta}_0,\hat{\beta}_1) = \sum_{i=1}^n (y_i - \overline{y})^2
    $$

  * The deviation from the outcome's mean predicted by the model, called explained deviation and also designated $ESS$:

    $$
        S_{\text{mod}}(\hat{\beta}_0,\hat{\beta}_1) = \sum_{i=1}^n (\hat{y}_i - \overline{y})^2
    $$

  * The deviation from the outcome's mean missed by the model, called unexplained deviation and also designated $RSS$, which corresponds to the residuals:

    $$
        \begin{align}
            S_{\text{res}}(\hat{\beta}_0,\hat{\beta}_1) &= \sum_{i=1}^n r_i^2 \\
             &= \sum_{i=1}^n (y_i - \hat{y}_i)^2 \\
        \end{align}
    $$

Then the coefficient of determination is:

$$
    R^2 = \frac{\text{explained variation in }y}{\text{total variation in }y} = \frac{S_{\text{mod}}}{S_{\text{tot}}} = 1 - \frac{S_{\text{res}}}{S_{\text{tot}}}
$$

With a simple linear regression, we also have $R^2 = r^2$, with $r$ the Pearson correlation coefficient.

In multiple linear regression, adding new predictors always increases the coefficient of determination, even just weakly. So adding many predictors with low predicting power can lead to a artificially high $R^2$. This becomes a problem when comparing models with different number of predictors. The adjusted coefficient of determination is the standard way to try to counteract that effect:

$$
    R^2_{\text{adj}} = 1 - \frac{S_{\text{res}}/\nu_{\text{res}}}{S_{\text{tot}}/\nu_{\text{tot}}}
$$

Where the two degrees of freedom are equal to:

$$
    \nu_{\text{res}} = n - m - 1, \qquad \nu_{\text{tot}} = n - 1
$$

(2:regression:residual_analysis)=
### Residual analysis

Residual analysis is an essential component of validating the results of a linear regression. The goal is to check that the assumption behind linear regression are indeed met. You will see in the next section that this can be done through summary statistics and hypothesis tests but, as usual, it is always valuable to visualize the results as well. Three main scatter plots are often used for this:
  * The predictor(s) against the residuals ({numref}`figure {number} <part-b_residual-analysis>`).
  * The estimated outcome against the residuals.
  * The estimated outcome against the true outcome from the data.

The goal is to identify any bias (i.e., the residuals' mean is different from 0), any heteroscedasticity (i.e., the residuals' variance is not constant), any correlation (i.e., either between residuals and predictors or between the residuals themselves, e.g., any non-linear relationship). This step can also help identify outliers that could perturb the analysis.

You have ```n``` = 12 monthly measurements of the height of a point on a glacier. The measurements are obtained from a satellite laser altimeter.

We will consider three different models, with the following observation equations:

- Model 1: constant velocity, $y_i = \beta_0 + \beta_1 t_i+\epsilon_i$
- Model 2: 8th order polynomial, $y_i =\beta_0 + \beta_1 t_i + \beta_2 t_i^2 +\ldots+ \beta_8 t_i^8 = \sum_{p=0}^8 \beta_p t_i^p +\epsilon_i$
- Model 3: constant velocity + annual signal, $y_i= \beta_0 + \beta_1 t_i + \beta_2 \cos \Big(\frac{2 \pi t_i}{12} \Big)+\epsilon_i$

````python
# times of observation [months]
t = np.arange(12)

# observed heights [m]
y = [102.4, 98.2, 97.5, 97.9, 99.7, 100.7, 98.3, 94.2,
                                     90.9, 86.1, 81.2, 76.9]

n = len(t)

X_1_ans = np.column_stack((np.ones(n), t))
X_2_ans = np.column_stack((X_1_ans, t**2, t**3, t**4,
                                    t**5, t**6, t**7, t**8))
X_3_ans = np.column_stack((np.ones(n), t, np.cos(2*np.pi*t/12)))

def lsqe_ans(y,X):
    '''Apply least squares estimation
    Input:
    y : vector with observations
    X : design matrix 
    '''
    # estimated parameters
    betahat = np.linalg.inv(X.T @ X) @ X.T @ y
    # adjusted observations
    yhat = X @ betahat
    # residuals
    ehat = y - yhat
    # squared norm of residuals
    eTe = ehat.T @ ehat
    
    return betahat, yhat, ehat, eTe

betahat_1_ans, yhat_1_ans, ehat_1_ans, eTe_1_ans = lsqe_ans(y,X_1_ans)
betahat_2_ans, yhat_2_ans, ehat_2_ans, eTe_2_ans = lsqe_ans(y,X_2_ans)
betahat_3_ans, yhat_3_ans, ehat_3_ans, eTe_3_ans = lsqe_ans(y,X_3_ans)
````

One of the models is *underfitting*. Try to reason what is meant by this and which model this applies to.

 ```{admonition} Solution
:class: tip, dropdown
Model 1 obviously does not capture the annual signal, so it is does not accurately describe the relation between observations and height change. This is referred to as underfitting.
```
       
One of the models is *overfitting*. Try to reason what is meant by this and which model this applies to.  

 ```{admonition} Solution
:class: tip, dropdown
Model 2 is overfitting - it results in very small residuals since we include a lot of parameters, but what is the physical interpretation? Including higher orders will result in an even better fit (ultimately the perfect fit, see below). A big problem is that most likely this model will not be able to predict the height change for the future. This is illustrated in the figure below, where we predict the height for the coming 2 months, and see how the polynomial model predicts a very unrealistic increase in height.
```

What would happen if you would fit a 11th order polynomial?

 ```{admonition} Solution
:class: tip, dropdown
This would result in a perfect fit: $\boldsymbol{X}$ becomes a full-rank square matrix, such that $\hat{\boldsymbol{\beta}} =\left(\boldsymbol{X}^T \boldsymbol{X}\right)^{-1} \boldsymbol{X}^T \boldsymbol{y}= \boldsymbol{X}^{-1}\boldsymbol{X}^{-T} \boldsymbol{X}^T \boldsymbol{y}=\boldsymbol{X}^{-1} \boldsymbol{y}$ and the residuals become 0 (show yourself).
```

How is redundancy defined?

 ```{admonition} Solution
:class: tip, dropdown
Redundancy is defined as the number of observations $n$ minus the number of unknown parameters $m$. 

From linear algebra we know that we can solve a linear system of equations with $m$ parameters, if we have at least $m$ independent equations. Therefore we call $n-m$ the redundancy, since it is the number of 'additional' observations compared to the minimum number required.
```

Do the residuals (see plot) behave as expected for each of the models?

 ```{admonition} Solution
:class: tip, dropdown
You would expect them to fluctuate randomly around zero if your model is correct, but especially for model 1 you can see a clear residual signal, which is an indication that the model is too simplistic.  
```
    
Discuss why the squared norm of residuals is quite different for the 3 models.

 ```{admonition} Solution
:class: tip, dropdown
The squared norm for model 1 is very large, due to the annual signal that is not accounted for. 

The squared norm for model 2 is very small, so it results in a very good fit (due to the high order of the polynomial). 

Model 3 results in a somewhat larger squared norm than model 2 - this is due to the additional parameters, which will result in a better fit (i.e., it allows to better capture the actual height variation over time).
```

````python
def plot_prediction():
    # create array with times
    t_pred = np.arange(14)
    n_pred = len(t_pred)

    # create the A-matrices for the given times
    X_pred_1 = np.column_stack((np.ones(n_pred), t_pred))
    X_pred_2 = np.column_stack((X_pred_1, t_pred**2, t_pred**3, t_pred**4, t_pred**5, t_pred**6, t_pred**7, t_pred**8))
    X_pred_3 = np.column_stack((np.ones(n_pred), t_pred, np.cos(2*np.pi*t_pred/12)))

    # calculate predicted heights
    model_1 = X_pred_1 @ betahat_1
    model_2 = X_pred_2 @ betahat_2
    model_3 = X_pred_3 @ betahat_3

    # plot the predicted heights bases on the three models
    plt.figure()
    plt.plot(t_pred, model_1, color='r', label='model 1')
    plt.plot(t_pred, model_2, color='c', label='model 2')
    plt.plot(t_pred, model_3, color='b', label='model 3')
    plt.xlim(-0.2, (n_pred-1)+0.2)
    plt.xlabel('time [months]')
    plt.ylabel('predicted height [meters]')
    plt.legend(loc='best')
    plt.show()
````

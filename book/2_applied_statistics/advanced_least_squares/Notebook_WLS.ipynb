:::{card} Explain the different results, in particular the 'locations' of the different lines with respect to each other.


 ```{admonition} Solution
:class: tip, dropdown 
The ordinary least squares fit ($W=I_n$) is 'in the middle' as expected, since all observations get equal weight. For case 2, the last 3 observations get a larger weight, so the line is pulled 'down' to result in smaller errors for those 3 observations. For case 3, the opposite effect occurs, since the two observations that get the largest weight are also 'above' the red line.
```

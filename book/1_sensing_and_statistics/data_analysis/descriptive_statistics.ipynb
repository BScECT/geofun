# Descriptive statistics

## What are descriptive statistics?

descriptive statistics vs descriptive statistics

  * *Univariate analysis* aims at describing the distribution of a single variable.
  * *Bivariate analysis* (and *multivariate analysis*) aims at describing the relationship between two variables (and more).

Exploratory Data Analysis

## Univariate analysis

### Measures of central tendency

Mean

Median

Mode

### Measures of dispersion

Variance

Standard deviation

Percentiles and quantiles

+ minimum and maximum

### Higher-order moments

Skewness

Kurtosis

## Bivariate analysis

### Measures of dependence

Covariance

Pearson correlation coefficient

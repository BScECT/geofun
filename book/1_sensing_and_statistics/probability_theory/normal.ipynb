# Table: Standard Normal Distribution

The table shows one-sided (**right-hand**) probabilities $\alpha$ as function of the value
$k_\alpha$, i.e. $\alpha = P(Z\geq k_\alpha)$.

To evaluate the table for $k_\alpha$: find the value up to the first decimal is in the first column, then find the column in the first row that most closely matches the second decimal.

Example: $\alpha$ = 0.0250 for $k_\alpha$ = 1.96.

You can also check with Python!

````python
from scipy.stats import norm
````

The cell below is set up to use interactively. To use it, click {fa}`rocket` --> {guilabel}`Live Code` on the top right corner of this screen and then wait until Python interaction is ready. The method `scipy.stats.norm` has already been imported as `norm`.

````python
alpha = 0.0250

k_alpha = norm.ppf(1 - alpha)

print(f"For alpha = {alpha:.4f}, k_alpha = {k_alpha:0.4f}.")
print(f"The probability in the upper (right-hand) tail is {alpha:.3f}.")
````

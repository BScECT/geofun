# Descriptive statistics

## What is descriptive statistics?

Descriptive statistics is a branch of statistics that aims to analyze and summarize a dataset. It does so by using measures (called summary statistics) and graphical representations of the data to convey as much information as possible in a simple and clear way. Those measures and representations can be split based on what they are analyzing:

  * *Univariate analysis* aims at describing the distribution of a single variable.
  * *Bivariate analysis* (and *multivariate analysis*) aims at describing the relationship between two variables (and more).

Contrary to inferential statistics, descriptive statistics does not aim at learning about the wider population from which the sample is drawn. So it often relies on empirical measures from the data, rather than estimating the true measures of the population using a model (e.g., of the underlying distribution) fitted on the data. Those empirical measures are computed directly from the observed data and typically do not require assuming a specific parametric model for the underlying distribution.

As such, descriptive statistics is a central component of exploratory data analysis (EDA): the process of exploring and analyzing the data directly to see what structures and information can be extracted from them, with a strong focus on data visualization. This approach was championed by the American statistician [John Tukey](https://www.theguardian.com/news/2000/aug/16/guardianobituaries1) in the 1960s and 1970s, who stressed the need to understand the data before going ahead with modeling.

## Summary statistics

For this section, let's assume that we have a sample of $n$ observed values $\{x_1, x_2, ..., x_n\}$ of a variable $X$ and $n$ observed values $\{y_1, y_2, ..., y_n\}$ of another variable $Y$.

#### The mean

The arithmetic mean is often simply called the mean or the average (but be aware that the geometrical and harmonic means also exist). It represents the center of mass of the distribution. Imagine 3D-printing the distribution as an object: it would balance perfectly on your finger if your finger were positioned at the mean. The empirical mean $\overline{x}$ of a sample is:

$$
    \overline{x} = \frac{1}{n}\sum_{i=1}^{n}x_i
$$

The symbol $\hat{\mu}_X$ can also be used: while $\mu_X$ usually represents the mean of the entire population, the hat symbol implies that this is an estimate of $\mu_X$. In settings where we have few, biased data, mean of the sample and mean of the population can differ a lot, but this is a problem for statistical inference, not descriptive statistics.

#### The median

The median is the middle value of the distribution, so the value that separates the lower half from the higher half. Its empirical value $\mathrm{med(X)}$ is:

$$
    \mathrm{Med}(X) = 
        \begin{cases}
          x_{\frac{n+1}{2}} & \text{if $n$ is odd}\\
          \frac{x_{\frac{n}{2}} + x_{\frac{n}{2}+1}}{2} & \text{if $n$ is even}\\
        \end{cases}
$$

Where the values are assumed to be sorted in increasing order. There is no universal symbol for the median, here we use $\mathrm{Med(X)}$, but $\mathrm{M}$ is sometimes used.

#### The mode

The mode is the most frequent value of the distribution, which can be defined as the value where the probability density function $f_{X}$ takes its maximum value:

$$
    \mathrm{Mod(X)} = \argmax_{x_i} \,f_{X}(x_i)
$$

Where $\mathrm{argmax}$ is the function returning the value $x_i$ where $f_{X}$ is maximal. Its empirical value can be determined from the empirical probability density function $\hat{f}_{X}$:

$$
    \hat{f}_{X}(x) = \frac{1}{n}\sum_{i=1}^{n}\mathbf{1}(x_i = x)
$$

Where $\mathbf{1}$ is the indicator function (equal to 1 when $x_i$ equals $x$, to 0 otherwise). There is also no universal symbol for the mode, here we use $\mathrm{Mod(X)}$, but $\mathrm{M_0}$ is sometimes used.

```{admonition} Tip: Which one to use?
:class: tip

The mean requires metric variables (which have valid distances between values), so it applies only to numerical variables. The median requires ordered variables (whose values can be ranked), so it applies to continuous, discrete, and ordinal variables. However, the formula above can only be used with continuous and discrete variables in the even case. For ordinal variables, we can only say that the median lies between $x_{\frac{n}{2}}$ and $x_{\frac{n}{2}+1}$. The mode has no specific requirements, so it applies to any variable type.

But the empirical value of the mode as defined above can make little sense with continuous variables: observing multiple times the exact same value can be unlikely and, if it happens, it might be far off the peak of the distribution, so far off the true mode of the underlying distribution. Another option is to turn to statistical inference and model the underlying distribution to estimate its mode. This is especially useful for multimodal distributions, which have several modes materializing as several peaks in the distribution.

So, in general, the empirical mode is used for categorical variables, the empirical mean or median for numerical ones. For discrete variables, the mean may not correspond to any actual valid value, and can therefore be difficult to interpret. For example, an average fertility rate of 1.49 children per parent is informative at the population level, but no individual has 1.49 children. 

The mean is also very sensitive to outliers, while the median is less influenced by these extreme observations, because it only depends on the middle value ({numref}`figure {number} <measures_central_tendency>`). A common example is income: a few very high earners can raise the mean considerably, whereas the median better reflects what a typical person earns. So the median is generally a more robust estimator of central tendency, especially for exploratory data analysis, except when outliers or skewness matter. An example is when looking at the errors of a model: if some errors are unusually high, we want to know, so the mean error is more valuable than the median error.
```

### Measures of dispersion

Measures of dispersion describe how spread out a distribution is, and you have already encountered the first two.

#### The variance

The variance is the mean of the squared deviation from the mean, so it quantifies the variation of the values of a distribution around the mean. Its empirical value $\mathrm{Var(X)}$ is:

$$
    \mathrm{Var(X)} = \frac{1}{n}\sum_{i=1}^{n}(x_i - \overline{x})^2
$$

$\mathrm{Var(X)}$ is the standard symbol for the variance of the random variable $X$, although $\sigma_X^2$ is sometimes used in reference to the link between the variance and our next measure, the standard deviation.

#### The standard deviation

The standard deviation also quantifies the amount of variation around the mean, defined as the square root of the variance.  Its empirical value $s_X$ is then:

$$
    s_X = \sqrt{\mathrm{Var(X)}} = \sqrt{\frac{1}{n}\sum_{i=1}^{n}(x_i - \overline{x})^2}
$$

The symbol $\hat{\sigma}_X$ can also be used: while $\sigma_X$ usually represents the standard deviation of the entire population, the hat symbol implies that this is an estimate of $\sigma_X$ (similarly to the mean).

In practice, when used with an estimate of the mean, this formula tends to underestimate the true standard deviation, especially with small sample sizes. So a correction factor of $\frac{n}{n-1}$, called Bessel's correction, is applied, leading to this updated formula:

$$
    s_X = \sqrt{\frac{1}{n-1}\sum_{i=1}^{n}(x_i - \overline{x})^2}
$$

You can check [here](https://en.wikipedia.org/wiki/Bessel%27s_correction) if you want to learn more about this correction and its derivation.

#### The quantiles

A quantile is a value such that a given proportion of the distribution lies at or below that value. It is then a generalization of the median that lets us assess the dispersion but also the asymmetry of the distribution. For proportions $\frac{1}{n} \le p \le \frac{n - 1}{n}$, one possible empirical definition is:

$$
    \begin{align}
        q_p &= x_{\lfloor h\rfloor} + (h - \lfloor h\rfloor)(x_{\lceil h\rceil} - x_{\lfloor h\rfloor})\\
        h &= (n + 1)p\\
    \end{align}
$$

Where the values are assumed to be sorted in increasing order, $\lfloor h\rfloor$ is the floor function that returns the largest integer smaller than or equal to $h$, and $\lceil h\rceil$ is the ceil function that returns the smallest integer larger than or equal to $h$. Quantiles are usually represented by $q_p$ or $Q_p$.

Quantiles are defined for proportions between 0 and 1, and they are called percentiles when $p$ is expressed as percentage. Five specific quantiles can be singled out:
  * The 0 quantile, or 0th percentile, which is the minimum; no data lie below it.
  * The 0.25 quantile, or 25th percentile, called the lower quartile; 25 % of the data lie below it.
  * The 0.5 quantile, or 50th percentile, which is the median; 50 % of the data lie below it.
  * The 0.75 quantile, or 75th percentile, called the upper quartile; 75 % of the data lie below it.
  * The 1 quantile, or 100th percentile, which is the maximum; all data lie at or below it.

The interquartile range (often abbreviated IQR) derives from those: it is the difference between upper and lower quartiles, so a direct measure of dispersion.

```{admonition} Tip: Which one to use?
:class: tip

Both the variance and the standard deviation are limited to numerical (continuous and discrete) variables. And they both lose their meaning for highly asymmetric distributions. The variance's unit is the squared unit of $X$, while the standard deviation has the same unit as $X$, making it easier to interpret. Quantiles are valid for continuous, discrete, and ordinal variables, and provides a finer view of the distribution's dispersion and asymmetry.

In general, the minimum, the lower quartile, the median, the upper quartile, and the maximum form a more robust set of measures to summarize and analyze a sample than the mean and the standard deviation.
```

### Measures of shape

Measures of shape describe the shape of the distribution with finer and finer details. Those are new, although they expand some concepts you have encountered before.

#### The skewness

The skewness quantifies the asymmetry of a distribution. While multiple formulas exist to define its empirical value $\mathrm{Skew(X)}$, you can use:

$$
    \mathrm{Skew(X)} = \frac{\frac{1}{n}\sum_{i=1}^{n}(x_i - \overline{x})^3}{s_X^3}
$$

For which the symbol $g_1$ is also used. It behaves as follows:
  * If $Skew(X) = 0$ then the distribution is symmetric, like the normal distribution.
  * If $Skew(X) > 0$ then the distribution has a longer tail to the right, so towards the high values.
  * If $Skew(X) < 0$ then the distribution has a longer tail to the left, so towards the low values.

#### The kurtosis

The kurtosis quantifies the heaviness of the tail of a distribution. Its empirical value $\mathrm{Kurt(X)}$ usually is:

$$
    \mathrm{Kurt(X)} = \frac{\frac{1}{n}\sum_{i=1}^{n}(x_i - \overline{x})^4}{s_X^4} - 3
$$

For which the symbol $g_2$ is also used. It behaves as follows:
  * If $Kurt(X) = 0$ then the distribution has the same tails as the normal distribution.
  * If $Kurt(X) > 0$ then the distribution has bigger tails than the normal distribution.
  * If $Kurt(X) < 0$ then the distribution has smaller tails than the normal distribution.

```{tip}

The moments are measures that characterize a distribution:
  * The mean is the first moment.
  * The variance is the second moment.
  * The skewness is the standardized version of the third moment.
  * The kurtosis is the standardized version of the fourth moment.

Higher-order moments also exist, and they characterize the shape of the distribution with more and more details. But they get more and more difficult to estimate, because they require larger and larger samples, and to interpret, so they are seldom used in exploratory data analysis.
```

### Measures of dependence

Measures of central tendency, dispersion, and shape are used for univariate analysis, i.e., to analyze a single variable. Measures of dependence on the other hand focus on bivariate analysis, i.e., to analyze the relationship between two variables.

#### The covariance

The covariance quantifies the joint variability between two variables. Its empirical value $\mathrm{Cov(X,Y)}$ is:

$$
    \mathrm{Cov(X,Y)} = \frac{1}{n}\sum_{i=1}^{n}(x_i - \overline{x})(y_i - \overline{y})
$$

$\mathrm{Cov(X,Y)}$ is the standard symbol for the covariance between the random variables $X$ and $Y$. It behaves as follows:
  * If $Cov(X,Y) = 0$ then there is no linear relationship between the two variables.
  * If $Cov(X,Y) > 0$ then $Y$ tends to increase when $X$ increases.
  * If $Cov(X,Y) < 0$ then $Y$ tends to decrease when $X$ increases.

:::{card} Definition

Formally, the covariance is defined as the expected value of the product of the deviation of the values of each variable with their respective expected values:

$$
    \mathrm{Cov(X,Y)} = \mathbb{E}[(X-\mathbb{E}[X])(Y-\mathbb{E}[Y])]
$$
:::

Amongst its properties, the covariance of a variable with itself is equal to the variance:

$$
    \mathrm{Cov(X,X)} = \mathrm{Var(X)}
$$

And it behaves similarly under linear transformations:

$$
    \mathrm{Cov(iX + j,kY + l)} = ik\mathrm{Cov(X,Y)}
$$

#### The Pearson correlation coefficient

The Pearson correlation coefficient quantifies the linear correlation between two variables. Its empirical value $\rho_{XY}$ is:

$$
    \rho_{XY} = \frac{Cov(X,Y)}{s_X s_Y}
$$

$\rho_{XY}$ is the standard symbol for the Pearson correlation coefficient between the random variables $X$ and $Y$. Since it is a normalized covariance, it is bounded between -1 and 1:

$$
    -1 \leq \rho_{XY} \leq 1
$$

And its behavior follows the same pattern ({numref}`figure {number} <correlation_coefficient>`):
  * If $\rho_{XY} = 0$ then $X$ and $Y$ are uncorrelated.
  * If $\rho_{XY} > 0$ then $X$ and $Y$ are positively correlated.
  * If $\rho_{XY} < 0$ then $X$ and $Y$ are negatively correlated.

```{admonition} Tip: Which one to use?
:class: tip

Both the covariance and the Pearson correlation coefficient are limited to numerical variables. And they are both limited to linear relationships, so $X$ and $Y$ can have a Pearson correlation coefficient of 0, and yet still show a clear non-linear relationship.

The covariance's unit is the product of the units of its two variables (e.g., if $X$ is in $m$ and Y in $m/s$, then $Cov(X,Y)$ is in $m^2/s$). This implies that the magnitude of the covariance depends on the magnitudes of $X$ and $Y$, which makes it difficult to interpret and to compare the covariances of multiple pairs of variables. The Pearson correlation coefficient on the other hand is normalized, so unitless, and has clear bounds, which makes it easy to interpret and compare.
```

## Graphical representations

As you have just seen, relying solely on summary statistics can lead to misanalyze the data, so visualizing them is essential. This is done through graphical representations called charts, plots, graphs, or diagrams in which the data are encoded with symbols. Similarly to summary statistics, different types of charts target specific types of analyses.

### Univariate analysis

#### Bar chart

A bar chart encodes the data as rectangular bars whose length is proportional to the values of the data. The bars can be displayed horizontally or vertically, and often do not touch each other. In general, each bar represents a category, so that bar charts are often used with categorical (nominal and ordinal) variables to compare the different categories ({numref}`figure {number} <part-a_descriptive-stats_bar-chart>`). They can also be used for discrete numerical variables. The bars can also be arranged from highest to lowest, to highlight the ranking between categories.

#### Histogram

A histogram also encodes the data as rectangular bars but to represent the distribution of a continuous variable. The bars are displayed vertically, and do touch each other because they discretize the continuous interval into contiguous bins. The bins of a histogram usually have equal width, meaning that all bars have the same width. But it is also possible to use bins of different widths within the same histogram to better highlight specific features of the data distribution. Each bar represents the number of data falling into the bin. It can also represent the density by dividing by the total number of data and the bin width.

The size or the number of bins is the key parameter of a histogram. While many formulas have been developed to select an optimal number of bins, there is in general no best value. In practice, it is better to look at multiple values, as they can highlight different features of a distribution ({numref}`figure {number} <histogram_bins>`):
  * A small number of bins better highlights the general shape of the distribution by smoothing out all the details.
  * A large number of bins better highlights the detailed variations of the distribution, including its potential multimodality and outliers.

#### Box plot

A box plot encodes the data as boxes expanded by lines (called whiskers) to represent and summarize the distribution of a continuous variable. The box and whiskers are built from the quartiles of the variable. The box itself is delimited by the lower and upper quartiles ($Q_1$ and $Q_3$), so it represents the interquartile range ($\text{IQR}$). A line inside of it represents the median. The whiskers represent the minimal value of the distribution within $Q_1 - 1.5\,\text{IQR}$ and the maximal value within $Q_3 + 1.5\,\text{IQR}$. If values exist beyond the $1.5\,\text{IQR}$ range, they are considered outliers and displayed as points.

Box plots aim at facilitating the analysis of the central tendency, dispersion, and asymmetry of the variable's distribution, and in the process facilitating the comparison of multiple groups or of multiple variables ({numref}`figure {number} <part-a_descriptive-stats_box-plot>`). This makes box plots particularly useful to quickly summarize and analyze multiple variables at once, but they also leave out a lot of details, so they do not replace histograms.

### Bi- and multi-variate analysis

#### Scatter plot

A scatter plot encodes the data of two variables as points displayed using a Cartesian coordinate system, with each dimension corresponding to a variable. The points themselves can use further encoding to display more variables: using different symbols (often called markers, e.g., circles, squares, triangles), using different sizes for those symbols, and using different colors, a scatter plot can encode up to 5 variables (but adding so much information can make the plot harder to decipher). It is also possible to display a scatter plot with three dimensions, so three variables, but it usually becomes harder to analyze.

Beside visualizing the relationship between variables ({numref}`figure {number} <part-a_descriptive-stats_scatter-plot>`), scatter plots help identify:
  * Extrema, i.e., extreme values that still follow the general pattern of the data.
  * Outliers, i.e., extreme values that fall outside the general pattern of the data.
  * Clusters, i.e., groups of data that follow specific patterns.

#### Pair plot

A pair plot combines histograms and scatter plots into a matrix or a grid to analyze more than two variables within a single chart. Each row of the grid corresponds to a variable, and the columns correspond to the same variables. The main diagonal then compares a variable with itself, so it displays the variables' histograms, while all the other elements compare two variables, so they display scatter plots ({numref}`figure {number} <part-a_descriptive-stats_pair-plot>`). The matrix is symmetric, so the lower and upper triangles contain the same information.

## Correlation vs. causation

In descriptive statistics, you analyze the correlation (sometimes called the association) between different variables. But always keep in mind that a strong correlation does not necessarily imply a cause-and-effect relationship. For instance, when it gets hot and sunny, we can observe that people eat more ice cream and have more sunburns, so there is a correlation between number of ice creams being eaten and number of sunburns, yet we cannot conclude that eating ice cream cause sunburns. The hidden variable here is weather (temperature and sunshine), which causes both ice cream consumption and sunburns.

A correlation without physical causes is called a spurious correlation, and you can find examples of spurious correlations [here](https://www.tylervigen.com/spurious-correlations). Whether a correlation is spurious or not matters in modeling: spurious correlations do not always hold in every conditions, and can lead to misunderstand the relationships in the data and to make poor predictions.
